```python
import numpy as np
import pandas as pd
import geopandas as gpd
import os.path as path
import utils
import matplotlib.pyplot as plt
```

```python
lau_df = gpd.read_file(path.join(utils.out_data_dir, "drought_days_lau.geojson"))
```

```python
nuts3_df = gpd.read_file(path.join(utils.out_data_dir, "nuts3_stats.geojson"))
# nuts3_df = gpd.read_file(path.join(utils.out_data_dir, "drought_days_nuts3.geojson"))
```

```python
max(lau_df["median_drought_days"].values), max(nuts3_df["median_drought_days"].values)
```

```python
fig, axs = plt.subplots(2, 1, sharex=True)
ax1, ax2 = axs
ax1.hist(nuts3_df["median_drought_days"].values, bins=365, range=(0, 365) )
ax1.set_xlabel("median drought days per year")
ax1.set_ylabel("Number of NUTS3 regions")

ax2.hist(lau_df["median_drought_days"].values, bins=365, range=(0, 365) )
ax2.set_xlabel("median drought days per year")
ax2.set_ylabel("Number of LAUs")
plt.show()
```
