```python
import os.path as path
import utils
import copy
import json
```

```python
median_drought_intervals = [25, 45, 65, 85, 105, 130, 155, 175, 200, 366];
median_drought_colors = [
    "#FFFFFF",
    "#FFE3AE",
    "#FFC655",
    "#FFAA00",
    "#FF7100",
    "#FF3355",
    "#C70021",
    "#901F32",
    "#6C2F39",
    "#452D31",
]

max_drought_intervals = [100, 175, 250, 300, 366];
max_drought_colors = [
    "#FFFFFF",
    "#FFE3AE",
    "#FFAA00",
    "#FF3355",
    "#901F32",
]


fallback_color = '"#808080"'
```

```python
colormap_template = '''
[
        "<",
        [
          "get",
          "{param_name}"
        ],
        {value}
      ],
    "{color}",'''

    
```

```python
nuts3_dict = {
        "id": "nuts3_stats",
        "type": "fill",
        "source": "nuts3_stats",
        "source-layer": "nuts3_stats",
        "paint": {
            "fill-opacity": [
                "case",
                ["boolean", ["feature-state", "hover"], False],
                0.25,
                0,
            ],
            "fill-color": [
                "case",
            ],
            "fill-outline-color": [
                "interpolate",
                ["linear"],32
                ["zoom"],
                0,
                "#fefefe00",
                100,
                "#fefefe00",
            ],
        },
       
    }
```

```python
nuts3_dict["paint"]["fill-color"]
```

```python
fill_col = ["case"]
for i in range(len(median_drought_intervals)):
    colormap_template = ["<", ["get", "median_drought_days"], median_drought_intervals[i]], median_drought_colors[i]
    fill_col.extend(colormap_template)

fill_col.append("#808080")
print(fill_col)
temp = copy.deepcopy(nuts3_dict)
temp["paint"]["fill-color"] = fill_col
# temp["id"] = "nuts3_stats_median"
with open(path.join(utils.out_data_dir, "median_drought_days" + "_nuts3_config.json"), "w+") as f:
    json.dump(fill_col, f, indent=4)
```

```python
fill_col = ["case"]
for i in range(len(max_drought_intervals)):
    colormap_template = ["<", ["get", "max_drought_days"], max_drought_intervals[i]], max_drought_colors[i]
    fill_col.extend(colormap_template)

fill_col.append("#808080")
print(fill_col)
temp = copy.deepcopy(nuts3_dict)
temp["paint"]["fill-color"] = fill_col
# temp["id"] = "nuts3_stats_max"

with open(path.join(utils.out_data_dir, "max_drought_days" + "_nuts3_config.json"), "w+") as f:
    json.dump(fill_col, f, indent=4)
```
